```python
import matplotlib.pyplot as plt
```

### Movies vs TV Shows Bar Chart

```python
df['type'].value_counts().plot(kind='bar')

plt.title('Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

### Top 10 Countries Chart

```python
df['country'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Countries by Netflix Content')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

### Top 10 Genres Visualization Chart

```python
df['listed_in'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

### Top 10 Content Ratings on Netflix

```python
df['rating'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Content Ratings on Netflix')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

### Netflix Content by Release Year

```python
yearly_content = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 6))
plt.plot(yearly_content.index, yearly_content.values)

plt.title('Netflix Content by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

### Movies vs TV Shows year-wise comparison

```python
content_by_year = df.groupby(['release_year', 'type']).size().unstack(fill_value=0)

content_by_year.tail(10)
```

### Line Chart

```python
content_by_year.plot(figsize=(12, 6))

plt.title('Movies vs TV Shows by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.legend(title='Type')
plt.grid(True)
plt.show()
```

### Duration Analysis

```python
movies = df[df['type'] == 'Movie'].copy()

movies['duration_min'] = movies['duration'].str.extract(r'(\d+)').astype(float)

movies[['title', 'duration', 'duration_min']].head(10)
```

### Duration visualization

```python
plt.figure(figsize=(10, 6))
plt.hist(movies['duration_min'].dropna(), bins=30)

plt.title('Distribution of Movie Duration')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')
plt.grid(True)
plt.show()
```

### Average Movie Duration

```python
movies['duration_min'].mean()
```

### Seasons convert to numeric 

```python
tv_shows = df[df['type'] == 'TV Show'].copy()

tv_shows['seasons'] = tv_shows['duration'].str.extract(r'(\d+)').astype(float)

tv_shows[['title', 'duration', 'seasons']].head(10)
```

### Most common number of seasons

```python
tv_shows['seasons'].value_counts().sort_index()
```

### Seasons chart

```python
tv_shows['seasons'].value_counts().sort_index().plot(kind='bar')

plt.title('Number of Seasons in Netflix TV Shows')
plt.xlabel('Number of Seasons')
plt.ylabel('Number of TV Shows')
plt.show()
```

### date_added convert to datetime

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

### Year-wise added content

```python
added_by_year = df['date_added'].dt.year.value_counts().sort_index()

added_by_year
```
